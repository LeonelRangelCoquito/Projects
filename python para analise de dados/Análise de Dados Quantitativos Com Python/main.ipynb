- DISTRIBUIÇÕES DE FREQUÊNCIA:
    1 Determinar o número de classes:
        k = raiz(n)
        n é o tamanho do conjunto de dados ou amostra
        ex: [2,3,1,5]
        n = 4

    2 Determinar a amplitude total AT
        é a diferença entre o maior e o menor valor no conjunto de dados 
    
    3 Determinar amplitude de classe AC
        AC = AT/k
    
    4 Contrução


```python
dados = [160, 165, 166, 168, 170, 170, 172, 174, 175, 175, 175, 178, 180, 180, 182, 183, 185, 185, 187, 188, 188, 190, 191, 195, 198, 200]
```

```python
# determinar numero de classes K
# k = raiz(n)
n = 0
for dado in dados:
    n += 1
k = round(n**0.5)
print(f'Soma dos N valores em dados é: {n}')
print(f'Número de classes K dado por raiz de n: {k}')

```
